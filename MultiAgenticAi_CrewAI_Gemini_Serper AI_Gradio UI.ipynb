```python
import os
import gradio as gr
from crewai import Agent, Task, Crew, LLM
from crewai_tools import SerperDevTool
```

```python
search_tool = SerperDevTool()


llm=LLM(
    model='gemini/gemini-2.5-flash',
    api_key=os.environ['GEMINI_API_KEY'],
    temperature=0.5,
)
def run_multi_agent(topic):
  teacher=Agent(
    role="Teacher",
    goal='Explain the concept clearly step by step',
    llm=llm,
    verbose=True,
    backstory='Experienced Teacher who explian the with expample'
    )
  researcher=Agent(
    role="Researcher",
    goal='find latest information',
    llm=llm,
    tools=[search_tool],
    verbose=True,
    backstory='expert reaching in latest information'
    )
  simplifier=Agent(
    role="Simplifier",
    goal='Making things easy',
    llm=llm,
    verbose=True,
    backstory='breack complex ideas into simple languae'
    )
  student=Agent(
    role="Student",
    goal='take notes',
    backstory='write simple notes like learner',
    llm=llm,
    verbose=True
    )
  examiner=Agent(
    role="Examiner",
    goal='create quetsions',
    llm=llm,
    verbose=True,
    backstory='test unsrstanding'
    )
  task1=Task(
    description=f'explain  {topic} in simple steps with example',
    expected_output='step-by-step by explanation',
    agent=teacher,
    )
  task2=Task(
    description=f'search find and import 3 points about {topic}',
    expected_output='3 clear points',
    agent=researcher
    )
  task3=Task(
    description=f'simplify the explanation of {topic}',
    expected_output='very simple explanation',
    agent=simplifier,
    )
  task4=Task(
    description=f'write short notes on {topic} like a student',
    expected_output='simple notes',
    agent=student,
  )
  task5=Task(
    description=f'create 5 questions about {topic}',
    expected_output='step by step explanation',
    agent=examiner,
    )
  crew = Crew(
    agents=[teacher, researcher, simplifier, student, examiner],
    tasks=[task1, task2, task3, task4, task5],
    verbose=False
    )

  result=crew.kickoff()

  return str(result)


# Define a custom theme
class CustomColorTheme(gr.themes.Base):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.name = "CustomColorTheme"
        self.background_fill_primary = "#F0FFF0"  # Light Green
        self.background_fill_secondary = "#FFFACD" # Lemon Chiffon (light yellow)
        self.button_primary_background_fill = "#FF6347" # Tomato (red)
        self.button_secondary_background_fill = "#ADFF2F" # GreenYellow


interface=gr.Interface(
    fn=run_multi_agent,
    inputs=gr.Textbox(label='Enter topic here',placeholder='Agentic AI VS AI Agent'),
    outputs=gr.TextArea(label='Multiagent output here'),
    title='Multi Agentic AI Application',
    description='Enter the topic here to result from get multi agent',
    theme=CustomColorTheme()
    )

if __name__ == '__main__':
  interface.launch()
```
